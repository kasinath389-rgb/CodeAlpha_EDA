## Task3 : Data Visualization
### Project : Restuarant Tips Data Visualization
**Objective:**
To create meaningful visualization from the restuarant tips dataset and identify useful insights.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
url ="https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
data = pd.read_csv(url)
display(data.head())
```

```python
customers = data["day"].value_counts()
plt.figure(figsize=(6,4))
plt.bar(customers.index, customers.values)
plt.title("Number of customers by Day")
plt.xlabel("Day")
plt.ylabel("Number of customers")
plt.show()
```

```python
gender = data["sex"].value_counts()

plt.figure(figsize=(6,6))
plt.pie(gender.values, labels=gender.index, autopct="%1.1f%%")
plt.title("Customer Gender Distribution")
plt.show()
```

```python
plt.figure(figsize=(6,4))
plt.scatter(data["total_bill"],data["tip"])
plt.title("Total Bill vs Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")
plt.show()
```

# Conclusion

1. The barchart shows that Saturday had the highest number of customers.
2. The pie chart displays the distribution of male and female customers.
3. The scatter polt shows a positive relatioship between total bill and tip, indicating that customers who spend more generally give higher tips.
4. Data visualiztion made it easier to identify customer patterns and trends.
